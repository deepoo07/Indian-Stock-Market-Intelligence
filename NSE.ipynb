```python
import yfinance as yf
import pandas as pd

stocks = [
    "RELIANCE.NS",
    "TCS.NS",
    "SBIN.NS",
    "INFY.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS",
    "ITC.NS"
]

all_data = []

for stock in stocks:

    data = yf.Ticker(stock).history(
        period="5y",
        interval="1d",
        auto_adjust=False
    )

    data.reset_index(inplace=True)

    data["Ticker"] = stock

    all_data.append(data)

    print(f"Downloaded: {stock}")

combined_data = pd.concat(all_data, ignore_index=True)

combined_data.head()
```

```python
combined_data.to_csv("Indian_Stock_Market_Data.csv", index=False)
```
